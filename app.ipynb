```python
import os
from dotenv import load_dotenv
from google import genai
import ipywidgets as widgets
from IPython.display import display, Markdown

print("All libraries imported successfully!")
```

```python
# Load API key from .env
load_dotenv()

api_key = os.getenv("GEMINI_API_KEY")

# Check API key
if not api_key:
    raise ValueError("Gemini API key not found! Check your .env file.")

# Initialize Gemini client
client = genai.Client(api_key=api_key)

print("Gemini connected successfully!")
```

```python
def ask_ai(prompt):

    response = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=prompt
    )

    return response.text


print("AI function created successfully!")
```

```python
answer = ask_ai("What is an LLM? Explain in simple words.")

print(answer)
```

```python
question = widgets.Text(
    placeholder="Ask Gemini anything...",
    description="You:",
    layout=widgets.Layout(width="75%")
)

send_button = widgets.Button(
    description="Send",
    button_style="primary"
)

chat_output = widgets.Output()

def send_message(_):
    prompt = question.value.strip()

    if not prompt:
        return

    question.value = ""

    with chat_output:
        display(Markdown(f"**You:** {prompt}"))

        try:
            answer = ask_ai(prompt)
            display(Markdown(f"**Gemini:**\n\n{answer}"))
        except Exception as e:
            print(f"Error: {e}")

send_button.on_click(send_message)

display(widgets.VBox([
    widgets.HBox([question, send_button]),
    chat_output
]))
```
